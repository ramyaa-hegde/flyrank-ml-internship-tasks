## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

url = (
    "https://raw.githubusercontent.com/flyrank-bih/"
    "flyrank-ml-internship-starter/main/"
    "data/raw/content_refresh_anonymized.csv"
)
df = pd.read_csv(url)

# Focus on pages with enough search visibility to make a CTR review meaningful.
visible_pages = df[df["impressions_90d"] >= 100].copy()

# Compare each visible page with the median CTR of pages in the same position tier.
visible_pages["expected_ctr_by_position_tier"] = (
    visible_pages.groupby("position_tier")["ctr"].transform("median")
)
visible_pages["ctr_gap_percentage_points"] = (
    visible_pages["expected_ctr_by_position_tier"] - visible_pages["ctr"]
).clip(lower=0)

under_capturing_pages = visible_pages[
    visible_pages["ctr_gap_percentage_points"] > 0
]

print("--- Starter-data evidence for Lane A ---")
print(f"1. Total pseudonymized content items: {len(df):,}")
print(f"2. Visible items with at least 100 impressions: {len(visible_pages):,}")
print(
    "3. Visible items below their position-tier median CTR: "
    f"{len(under_capturing_pages):,}"
)
print(f"\nAverage CTR across all items: {df['ctr'].mean():.2f}%")
```
